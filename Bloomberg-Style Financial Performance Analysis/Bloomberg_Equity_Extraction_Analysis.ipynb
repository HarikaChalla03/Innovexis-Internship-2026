#### Data Cleaning Functions

````python
# ============================================================
# DATA CLEANING FUNCTIONS
# ============================================================

def clean_label(value):
    """
    Standardize text labels and remove footnote markers.
    """

    if pd.isna(value):
        return np.nan

    value = str(value).strip()

    value = re.sub(
        r"\(\d+\)",
        "",
        value
    )

    value = re.sub(
        r"\[\d+\]",
        "",
        value
    )

    value = re.sub(
        r"\s+",
        " ",
        value
    )

    return value.strip()


def clean_financial_value(value):
    """
    Convert Bloomberg-style financial values to numeric.
    """

    if pd.isna(value):
        return np.nan

    value = str(value).strip()

    missing_values = {
        "",
        "-",
        "—",
        "–",
        "NM",
        "N/M",
        "N/A",
        "NA",
        "#N/A",
        "#NA",
        "NULL",
        "NONE"
    }

    if value.upper() in missing_values:
        return np.nan

    value = value.replace(",", "")
    value = value.replace(" ", "")
    value = value.replace("%", "")

    if value.startswith("(") and value.endswith(")"):
        value = "-" + value[1:-1]

    value = re.sub(
        r"[^0-9.\-]",
        "",
        value
    )

    if value in {"", "-", ".", "-."}:
        return np.nan

    try:
        return float(value)

    except ValueError:
        return np.nan


def validate_reconciliation(
    df,
    components,
    total_label,
    group_column=None,
    tolerance=0.01
):
    """
    Validate that component values reconcile to reported total.
    """

    results = []

    years = sorted(
        df["fiscal_year"].dropna().unique()
    )

    for year in years:

        year_data = df[
            df["fiscal_year"] == year
        ]

        component_data = year_data[
            year_data["item"].isin(components)
        ]

        total_data = year_data[
            year_data["item"] == total_label
        ]

        component_sum = component_data["value"].sum()

        if len(total_data) > 0:
            reported_total = total_data["value"].iloc[0]
        else:
            reported_total = np.nan

        difference = (
            component_sum - reported_total
            if pd.notna(reported_total)
            else np.nan
        )

        status = (
            "PASS"
            if pd.notna(difference)
            and abs(difference) <= tolerance
            else "REVIEW_REQUIRED"
        )

        results.append({
            "Fiscal Year": int(year),
            "Component Sum": component_sum,
            "Reported Total": reported_total,
            "Difference": difference,
            "Status": status
        })

    return pd.DataFrame(results)
````

Metric Classification

````python
# ============================================================
# P&L — METRIC CLASSIFICATION
# ============================================================

percentage_metrics = {
    "Revenue Growth Y/Y",
    "Gross Margin %",
    "Operating Margin %",
    "Net Margin %",
    "Expense / Sales",
    "Effective Tax Rate %"
}

per_share_metrics = {
    "Basic EPS",
    "Diluted EPS",
    "Dividend Per Share (DPS)"
}

share_metrics = {
    "Wtd. Avg Shares - Basic (mn)",
    "Wtd. Avg Shares - Diluted (mn)"
}


def classify_metric(label):

    if label in percentage_metrics:
        return "Percentage"

    if label in per_share_metrics:
        return "Per Share"

    if label in share_metrics:
        return "Shares"

    return "Financial Value"


pl["metric_type"] = pl[
    "line_item"
].apply(
    classify_metric
)

display(
    pl[
        ["line_item", "metric_type"]
    ]
)
````

Gross Profit Validation

````python
# ============================================================
# P&L — GROSS PROFIT RECONCILIATION
# ============================================================

gross_profit_rows = []

for year in FISCAL_YEARS:

    revenue = pl.loc[
        pl["line_item"].eq("Total Net Revenue"),
        f"FY_{year}"
    ].iloc[0]

    cogs = pl.loc[
        pl["line_item"].eq("Cost of Goods Sold"),
        f"FY_{year}"
    ].iloc[0]

    reported_gp = pl.loc[
        pl["line_item"].eq("GROSS PROFIT"),
        f"FY_{year}"
    ].iloc[0]

    calculated_gp = revenue + cogs

    difference = calculated_gp - reported_gp

    gross_profit_rows.append({
        "Fiscal Year": year,
        "Revenue": revenue,
        "COGS": cogs,
        "Calculated Gross Profit": calculated_gp,
        "Reported Gross Profit": reported_gp,
        "Difference": difference,
        "Status": (
            "PASS"
            if abs(difference) <= 0.01
            else "REVIEW_REQUIRED"
        )
    })

gross_profit_validation = pd.DataFrame(
    gross_profit_rows
)

display(gross_profit_validation)
````

Gross Margin Validation

````python
gross_margin_rows = []

for year in FISCAL_YEARS:

    revenue = pl.loc[
        pl["line_item"].eq("Total Net Revenue"),
        f"FY_{year}"
    ].iloc[0]

    gross_profit = pl.loc[
        pl["line_item"].eq("GROSS PROFIT"),
        f"FY_{year}"
    ].iloc[0]

    reported_margin = pl.loc[
        pl["line_item"].eq("Gross Margin %"),
        f"FY_{year}"
    ].iloc[0]

    calculated_margin = (gross_profit / revenue) * 100

    difference = calculated_margin - reported_margin

    status = (
        "PASS"
        if abs(difference) <= 0.1
        else "REVIEW_REQUIRED"
    )

    gross_margin_rows.append({
        "Fiscal Year": year,
        "Calculated Margin": round(calculated_margin, 3),
        "Reported Margin": round(reported_margin, 3),
        "Difference": round(difference, 3),
        "Status": status
    })

gross_margin_validation = pd.DataFrame(gross_margin_rows)

display(gross_margin_validation)
````

EBIT Validation

````python
# ============================================================
# P&L — EBIT VALIDATION
# ============================================================

ebit_rows = []

for year in FISCAL_YEARS:

    gp = pl.loc[
        pl["line_item"].eq("GROSS PROFIT"),
        f"FY_{year}"
    ].iloc[0]

    other_income = pl.loc[
        pl["line_item"].eq("Royalty and Commission Income"),
        f"FY_{year}"
    ].iloc[0]

    other_operating_income = pl.loc[
        pl["line_item"].eq("Other Operating Income"),
        f"FY_{year}"
    ].iloc[0]

    total_opex = pl.loc[
        pl["line_item"].eq("Total Other Operating Exp."),
        f"FY_{year}"
    ].iloc[0]

    impairment_values = pl.loc[
        pl["line_item"].eq("Goodwill / Intangible Impairment"),
        f"FY_{year}"
    ]

    impairment = (
        impairment_values.iloc[0]
        if not impairment_values.empty
        else 0
    )

    if pd.isna(impairment):
        impairment = 0

    calculated_ebit = (
        gp
        + other_income
        + other_operating_income
        + total_opex
        + impairment
    )

    reported_ebit = pl.loc[
        pl["line_item"].eq("OPERATING PROFIT (EBIT)"),
        f"FY_{year}"
    ].iloc[0]

    difference = calculated_ebit - reported_ebit

    ebit_rows.append({
        "Fiscal Year": year,
        "Calculated EBIT": calculated_ebit,
        "Reported EBIT": reported_ebit,
        "Difference": difference,
        "Status": (
            "PASS"
            if abs(difference) <= 0.01
            else "REVIEW_REQUIRED"
        )
    })

ebit_validation = pd.DataFrame(ebit_rows)

display(ebit_validation)
````

EPS Validation

````python
# ============================================================
# P&L — EPS VALIDATION
# ============================================================

eps_rows = []

for year in FISCAL_YEARS:

    net_income = pl.loc[
        pl["line_item"].eq(
            "NET INCOME ATTRIB. TO SHAREHOLDERS"
        ),
        f"FY_{year}"
    ].iloc[0]

    shares = pl.loc[
        pl["line_item"].eq(
            "Wtd. Avg Shares - Basic (mn)"
        ),
        f"FY_{year}"
    ].iloc[0]

    reported_eps = pl.loc[
        pl["line_item"].eq("Basic EPS"),
        f"FY_{year}"
    ].iloc[0]

    calculated_eps = (
        net_income / shares
        if shares != 0
        else np.nan
    )

    difference = (
        calculated_eps - reported_eps
    )

    eps_rows.append({
        "Fiscal Year": year,
        "Calculated EPS": calculated_eps,
        "Reported EPS": reported_eps,
        "Difference": difference,
        "Status": (
            "PASS"
            if abs(difference) <= 0.05
            else "REVIEW_REQUIRED"
        )
    })

eps_validation = pd.DataFrame(
    eps_rows
)

display(eps_validation)
````

````python
# ============================================================
# REVENUE BY GEOGRAPHY — FIND YEAR ROW
# ============================================================

geo_year_row = None

for i in range(
    min(20, len(geo))
):

    row_text = " ".join(
        geo.iloc[i]
        .astype(str)
        .tolist()
    )

    if all(
        str(year) in row_text
        for year in [2016, 2017, 2018]
    ):

        geo_year_row = i
        break

print(
    "Geography year row:",
    geo_year_row
)
````

Create Geography Table

````python
# ============================================================
# REVENUE BY GEOGRAPHY — CREATE CLEAN TABLE
# ============================================================

if geo_year_row is None:

    raise ValueError(
        "Geography year row could not be detected."
    )

geo_columns = []

for i, value in enumerate(
    geo.iloc[geo_year_row]
):

    if pd.isna(value):

        geo_columns.append(
            f"column_{i}"
        )

    else:

        year_match = re.search(
            r"(\d{4})",
            str(value)
        )

        if year_match:

            geo_columns.append(
                f"FY_{year_match.group(1)}"
            )

        else:

            geo_columns.append(
                clean_label(value)
            )

geo.columns = geo_columns

geo = geo.iloc[
    geo_year_row + 1:
].copy()

geo = geo.dropna(
    how="all"
)

geo = geo.reset_index(
    drop=True
)

display(geo.head(25))
````

Standardize Geography Columns

````python
# ============================================================
# REVENUE BY GEOGRAPHY — STANDARDIZE COLUMNS
# ============================================================

geo.columns = [
    clean_label(column)
    for column in geo.columns
]

geo_label_column = geo.columns[0]

for column in geo.columns[1:]:

    if str(column).startswith("FY_"):

        geo[column] = geo[column].apply(
            clean_financial_value
        )

geo = geo[
    geo[geo_label_column].notna()
].copy()

geo = geo[
    geo[geo_label_column].astype(str).str.strip() != ""
].copy()

geo = geo.reset_index(
    drop=True
)

print("Geography columns:")
print(geo.columns.tolist())

display(geo.head(25))
````

Geography Region Mapping

````python
# ============================================================
# REVENUE BY GEOGRAPHY — REGION MAPPING
# ============================================================

def map_region(value):

    if pd.isna(value):
        return np.nan

    text = str(value).strip().lower()

    region_map = {

        "europe": "Europe",
        "north america": "North America",
        "asia": "Asia",
        "latin america": "Latin America",
        "emea": "EMEA",
        "greater china": "Greater China",
        "north america": "North America",
        "asia pacific": "Asia Pacific"
    }

    for key, region in region_map.items():

        if key in text:
            return region

    return "Other"


geo["region"] = geo[
    geo_label_column
].apply(map_region)

geo = geo.rename(
    columns={
        geo_label_column: "geography"
    }
)

display(geo.head(25))
````

Geography Long Format

````python
# ============================================================
# REVENUE BY GEOGRAPHY — LONG FORMAT
# ============================================================

geo_year_columns = [
    column
    for column in geo.columns
    if str(column).startswith("FY_")
]

geo_long = geo.melt(
    id_vars=[
        "geography",
        "region"
    ],
    value_vars=geo_year_columns,
    var_name="fiscal_year",
    value_name="revenue"
)

geo_long["fiscal_year"] = (
    geo_long["fiscal_year"]
    .str.replace("FY_", "")
    .astype(int)
)

geo_long["company"] = "Adidas AG"
geo_long["ticker"] = "ADS GY"
geo_long["currency"] = "EUR"
geo_long["unit"] = "Millions"

geo_long = geo_long[
    [
        "company",
        "ticker",
        "fiscal_year",
        "geography",
        "region",
        "revenue",
        "currency",
        "unit"
    ]
]

display(geo_long.head(25))
````

````python
# ============================================================
# OPEX DETAIL — FIND YEAR ROW
# ============================================================

opex_year_row = None

for i in range(
    min(20, len(opex))
):

    row_text = " ".join(
        opex.iloc[i]
        .astype(str)
        .tolist()
    )

    if all(
        str(year) in row_text
        for year in [2016, 2017, 2018]
    ):

        opex_year_row = i
        break

print(
    "OPEX year row:",
    opex_year_row
)
````

Create OPEX Table

````python
# ============================================================
# OPEX DETAIL — CREATE CLEAN TABLE
# ============================================================

if opex_year_row is None:

    raise ValueError(
        "OPEX year row could not be detected."
    )

opex_columns = []

for i, value in enumerate(
    opex.iloc[opex_year_row]
):

    if pd.isna(value):

        opex_columns.append(
            f"column_{i}"
        )

    else:

        year_match = re.search(
            r"(\d{4})",
            str(value)
        )

        if year_match:

            opex_columns.append(
                f"FY_{year_match.group(1)}"
            )

        else:

            opex_columns.append(
                clean_label(value)
            )

opex.columns = opex_columns

opex = opex.iloc[
    opex_year_row + 1:
].copy()

opex = opex.dropna(
    how="all"
)

opex = opex.reset_index(
    drop=True
)

display(opex.head(25))
````

Numeric Conversion

````python
# ============================================================
# OPEX DETAIL — NUMERIC CONVERSION
# ============================================================

opex.columns = [
    clean_label(column)
    for column in opex.columns
]

opex_label_column = opex.columns[0]

opex_year_columns = [
    column
    for column in opex.columns
    if str(column).startswith("FY_")
]

for column in opex_year_columns:

    opex[column] = opex[column].apply(
        clean_financial_value
    )

opex = opex[
    opex[opex_label_column].notna()
].copy()

opex = opex[
    opex[opex_label_column].astype(str).str.strip() != ""
].copy()

opex = opex.rename(
    columns={
        opex_label_column: "expense_item"
    }
)

display(opex.head(25))
````

OPEX Long Format

````python
# ============================================================
# OPEX DETAIL — LONG FORMAT
# ============================================================

opex_long = opex.melt(
    id_vars=[
        "expense_item"
    ],
    value_vars=opex_year_columns,
    var_name="fiscal_year",
    value_name="expense"
)

opex_long["fiscal_year"] = (
    opex_long["fiscal_year"]
    .str.replace("FY_", "")
    .astype(int)
)

opex_long["company"] = "Adidas AG"
opex_long["ticker"] = "ADS GY"
opex_long["currency"] = "EUR"
opex_long["unit"] = "Millions"

opex_long = opex_long[
    [
        "company",
        "ticker",
        "fiscal_year",
        "expense_item",
        "expense",
        "currency",
        "unit"
    ]
]

display(opex_long.head(25))
````

OPEX Validation

````python
opex_validation_rows = []

component_items = [
    "Marketing and point-of-sale",
    "Distribution and selling",
    "General and administration",
    "Sundry expenses",
    "Goodwill / intangible impairment"
]

for year in FISCAL_YEARS:

    year_data = opex_long[
        opex_long["fiscal_year"].eq(year)
    ].copy()

    component_data = year_data[
        year_data["expense_item"].isin(component_items)
        & (year_data["expense"] > 100)
    ]

    total_data = year_data[
        year_data["expense_item"].eq("Total operating expenses")
        & (year_data["expense"] > 100)
    ]

    component_sum = component_data["expense"].sum()

    reported_total = (
        total_data["expense"].iloc[0]
        if len(total_data) > 0
        else np.nan
    )

    difference = component_sum - reported_total

    status = (
        "PASS"
        if pd.notna(difference) and abs(difference) <= 0.01
        else "REVIEW_REQUIRED"
    )

    opex_validation_rows.append({
        "Fiscal Year": year,
        "Component Sum": round(component_sum, 2),
        "Reported Total": round(reported_total, 2)
        if pd.notna(reported_total) else np.nan,
        "Difference": round(difference, 2)
        if pd.notna(difference) else np.nan,
        "Status": status
    })

opex_validation = pd.DataFrame(opex_validation_rows)

display(opex_validation)
````
