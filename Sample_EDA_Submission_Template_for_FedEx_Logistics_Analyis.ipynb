````python
# DATE COLUMN ANALYSIS
 # DATE STANDARDIZATION & VALIDATION

import pandas as pd
import numpy as np


# ============================================================================
# STEP 1: Replace business text with NaN
# ============================================================================

cols = ['PQ First Sent to Client Date', 'PO Sent to Vendor Date', 'Scheduled Delivery Date', 'Delivered to Client Date', 'Delivery Recorded Date']

# Replace text with NaN
for text in ['Pre-PQ Process', 'Date Not Captured', 'N/A-FROM RDC']:
    for col in cols:
        df[col] = df[col].replace(text, np.nan)

# ============================================================================
# STEP 2: Parse with flexible multi-format parser
# ============================================================================
def parse_mixed_dates(date_val):
    if pd.isnull(date_val):
        return pd.NaT

    date_str = str(date_val).strip()

    # Try all common formats
    formats = [
        '%d-%b-%y', '%d-%b-%Y',      # 02-Jun-06, 02-Jun-2006
        '%d/%m/%y', '%d/%m/%Y',      # 02/06/06, 02/06/2006
        '%d-%m-%y', '%d-%m-%Y',      # 02-06-06, 02-06-2006
        '%m/%d/%Y', '%m/%d/%y',      # 06/02/2006, 06/02/06
        '%Y-%m-%d', '%Y/%m/%d',      # 2006-06-02, 2006/06/02
        '%d %b %Y', '%d %b %y',      # 02 Jun 2006, 02 Jun 06
        '%B %d, %Y', '%b %d, %Y',    # June 02, 2006, Jun 02, 2006
    ]

    for fmt in formats:
        try:
            return pd.to_datetime(date_str, format=fmt)
        except:
            continue

    return pd.NaT

# Apply to ALL columns
for col in cols:
    df[col] = df[col].apply(parse_mixed_dates)


print("Data types:")
print(df[cols].dtypes)

print("\nNon-null counts:")
for col in cols:
    print(f"  {col}: {df[col].notna().sum()} non-null")

print("\nSample dates:")
print(df[cols].head(5))
````

````python
#Standardization of Weight and Freight Cost Columns
df['Weight (Kilograms)'] = pd.to_numeric(
    df['Weight (Kilograms)'], errors='coerce'
)

df['Freight Cost (USD)'] = pd.to_numeric(
    df['Freight Cost (USD)'], errors='coerce'
)
````

# OUTLIER DETECTION (IQR Method)

````python
# OUTLIER DETECTION (IQR Method)

numeric_cols = [
    'Unit of Measure (Per Pack)',
    'Line Item Quantity',
    'Line Item Value',
    'Pack Price',
    'Unit Price',
    'Line Item Insurance (USD)',
    'Weight (Kilograms)',
    'Freight Cost (USD)'
]

#Count Outliers
outlier_summary = {}

total_rows = len(df)

for col in numeric_cols:
    clean_series = df[col].dropna()

    Q1 = clean_series.quantile(0.25)
    Q3 = clean_series.quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = clean_series[
        (clean_series < lower_bound) |
        (clean_series > upper_bound)
    ]

    outlier_count = len(outliers)
    outlier_percent = (outlier_count / total_rows) * 100

    outlier_summary[col] = {
        "Outliers": outlier_count,
        "Percentage": round(outlier_percent, 2)
    }

print(outlier_summary)

#Convert to Table
outlier_df = pd.DataFrame(outlier_summary).T
print(outlier_df)
````

I used the IQR method to identify statistical outliers across shipment quantity, pricing, weight, and freight cost. Since supply chain datasets naturally contain legitimate high-volume and high-cost shipments, I avoided blindly removing outliers. Instead, I retained business-valid extremes and created anomaly flags for delivery delays, weight, and freight cost to support root-cause analysis.

KPI 8 — Freight Outlier Distribution

````python
Q1 = clean_df['Freight Cost (USD)'].quantile(0.25)
Q3 = clean_df['Freight Cost (USD)'].quantile(0.75)

IQR = Q3 - Q1
upper = Q3 + 1.5 * IQR

clean_df['Freight_Outlier_Flag'] = np.where(
    clean_df['Freight Cost (USD)'] > upper,
    'Outlier',
    'Normal'
)
````

````python
outlier_dist = clean_df['Freight_Outlier_Flag'].value_counts()
print(outlier_dist)
````
