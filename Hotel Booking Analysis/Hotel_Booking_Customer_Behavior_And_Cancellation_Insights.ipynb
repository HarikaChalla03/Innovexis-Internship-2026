### Feature Engineering

````python
# Create new feature for total stay nights by summing up the stays in weekend nights and stays in weekdays nights

df["total_stay_nights"] = (
    df["stays_in_weekend_nights"] + df["stays_in_weekdays_nights"]
)
````

````python
# Create new feature for total guests by summing up the number of adults, children, and babies
df["total_guests"] = (
    df["adults"] + df["children"] +df["babies"]
)
````

````python
# Create new feature for arrival season based on the arrival month
season_map = {
    "January": "Winter", "February": "Winter", "December": "Winter",
    "March": "Spring", "April": "Spring", "May": "Spring",
    "June": "Summer", "July": "Summer", "August": "Summer",
    "September": "Autumn", "October": "Autumn", "November": "Autumn"
}

df["arrival_season"] = df["arrival_date_month"].map(season_map)
````

````python
# Create new feature for booking status based on the is_canceled column
df["booking_status"] = df["is_canceled"].map({
    0: "Completed",
    1: "Canceled"
})
````

#### Chart - 12 : Which hotel generates higher estimated room revenue from completed bookings?

Formula

Estimated Room Revenue = ADR × Total Stay Nights

````python
# Chart - 12 visualization code

df["estimated_room_revenue"] = (
    df["adr"] * df["total_stay_nights"]
)

# Comparing estimated room revenue by hotel type
revenue_by_hotel = (
    df.groupby("hotel")["estimated_room_revenue"]
    .sum()
    .sort_values(ascending=False)
)

print(revenue_by_hotel)

#Visualizing the estimated room revenue by hotel type
plt.figure(figsize=(8, 5))

sns.barplot(
    x=revenue_by_hotel.index,
    y=revenue_by_hotel.values
)

plt.title("Estimated Room Revenue by Hotel Type")
plt.xlabel("Hotel Type")
plt.ylabel("Estimated Room Revenue")

plt.show()
````
